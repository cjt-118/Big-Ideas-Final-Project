The metric I am measuring to determine the best neighborhood to live in is violent crime.

I will be using [](https://crimegrade.org) to get the data.

The graph below displays violent crimes per 1000 people in each neighborhood.

```python
import pandas as pd
import matplotlib.pyplot as plt

crimes = pd.read_csv("crime.csv")
crimes.columns = crimes.columns.str.strip()

counts = crimes[['Neighborhood', 'Crimerate']].copy()
counts = counts.sort_values(by='Crimerate', ascending=False)

plt.figure(figsize=(10, 4))
plt.barh(counts['Neighborhood'], counts['Crimerate'])
plt.title("Violent Crimes per 1000 People per Neighborhood")
plt.xlabel("Violent Crimes per 1000 People")
plt.ylabel("Neighborhood")
plt.tight_layout()
plt.show()
```

Using this data, Cranberry Township wins lowest crime rate with only 1.5 violent crimes per 1000 people. Meanwhile oakland where we are takes last place with 4.5 violent crimes per 1000 people. I have no favorite neighborhood as I have not been here for very long however it is vastly safer than oakland where I currently live.
